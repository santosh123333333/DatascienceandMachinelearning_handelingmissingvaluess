```python

import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer


print("Downloading dataset...")
url = "hf://datasets/t22000t/house-prices-tabular/train.csv"
train_df = pd.read_csv(url)


# (We take the last 100 rows of the train data)
test_df = train_df.tail(100).copy()
train_df = train_df.iloc[:-100].copy()
# Remove SalePrice from test data
test_df = test_df.drop(columns=["SalePrice"])

print(f"Train shape: {train_df.shape}")
print(f"Test shape: {test_df.shape}")
print("\nSuccessfully loaded both dataframes!")
```

```python
# display few dataset
test_df.head()
```

```python
X_train=train_df.drop(columns="SalePrice")
y_train=train_df["SalePrice"]


print(f"X_Train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
```

```python
print(train_df.columns)
```

```python
# Handeling Numerical missing values Imputation
# Select numerical columns
numerical_cols = X_train.select_dtypes(include=["int64", "float64"]).columns
numerical_cols
```

```python
# checking the missing values
X_train[numerical_cols].isnull().sum()
```

```python
imputer_mean=SimpleImputer(strategy='mean')
```

```python
imputer_mean.fit(X_train[numerical_cols])
```

```python
imputer_mean.statistics_
```

```python
imputer_mean.transform(X_train[numerical_cols])
```

```python
X_train[numerical_cols]=imputer_mean.transform(X_train[numerical_cols])
test_df[numerical_cols]=imputer_mean.transform(test_df[numerical_cols])

```

```python
X_train[numerical_cols].isnull().sum()
```

```python
test_df[numerical_cols].isnull().sum()
                                
```

# Handeling Categorical variable
Categorical Missing Value Imputation



```python
categorical_cols = X_train.select_dtypes(include=["object", "category"]).columns
categorical_cols
```

```python
X_train[categorical_cols].isnull().sum()
```

```python
imputer_mostfreq=SimpleImputer(strategy='most_frequent')
imputer_mostfreq
```

```python
imputer_mostfreq.fit(X_train[categorical_cols])
```

```python
imputer_mostfreq.statistics_
```

```python
# Transform the categorical columns using the fitted imputer
imputer_mostfreq.transform(X_train[categorical_cols])
```

```python
X_train[categorical_cols]=imputer_mostfreq.transform(X_train[categorical_cols])
test_df[categorical_cols]=imputer_mostfreq.transform(test_df[categorical_cols])
```

```python
X_train[categorical_cols].isnull().sum()
```

```python
test_df[categorical_cols].isnull().sum()
```
